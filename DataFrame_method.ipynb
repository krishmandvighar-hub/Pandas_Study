```python
import pandas as pd

# Create two DataFrames
df1 = pd.DataFrame({'A': [1, 2]}, index=['x', 'y'])
df2 = pd.DataFrame({'A': [3, 4]}, index=['y', 'z'])

# Align them
a1, a2 = df1.align(df2, join='outer', fill_value=0)#align inner,right,left

# Add after alignment
result = a1 + a2

print("Aligned df1:\n", a1)
print("\nAligned df2:\n", a2)
print("\nResult:\n", result)
```

```python
import pandas as pd

def calc_sum(x):
  return x.sum()

data = {
  "x": [50, 40, 30],
  "y": [300, 1112, 42]
}

df = pd.DataFrame(data)

x = df.apply(calc_sum)

print(x)
```

```python
import pandas as pd

df1 = pd.DataFrame([[10, 9], [5, 4]]) 
df2 = pd.DataFrame([[9, 7], [4, 8]]) 

def myfunc(a, b) :
  print(a,"\nrtyhj")
  print(b,"\n")
  if (a.sum() > b.sum()):
    return a
  else:
    return b

print(df1.combine(df2, myfunc))
```
