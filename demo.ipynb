```python
import pandas as pd

df = pd.read_csv("data.csv")

# Convert salary column to numeric (fixes empty/invalid values)
df["salary"] = pd.to_numeric(df["salary"], errors="coerce")

# Calculate mean salary (ignores NaN automatically)
mean_salary = df["salary"].mean()

# Fill missing salaries with mean
df["salary"].fillna(mean_salary, inplace=True)

print(df.to_string())
```

```python
import pandas as pd

df = pd.read_csv("data.csv")

# Convert salary to numeric (important step)
df["salary"] = pd.to_numeric(df["salary"], errors="coerce")

# Calculate median
x = df["salary"].median()

# Fill missing values with median
df["salary"].fillna(x, inplace=True)

print(df.to_string())
```

```python
import pandas as pd

df = pd.read_csv('data.csv')

df['Date'] = pd.to_datetime(df['Date'], format='mixed') #proper Format

df.dropna(subset=['Date'], inplace = True)#remove NAN values

print(df.to_string())
```

```python
import pandas as pd

df = pd.read_csv('data.csv')

for x in df.index:
  if df.loc[x, "Duration"] > 120:#Any condition we can apply
    df.loc[x, "Duration"] = 120 #replacing method 

print(df.to_string())
```

```python
#delete rows by drop() method with for loop and if condition
import pandas as pd

df = pd.read_csv('data.csv')

for x in df.index:
  if df.loc[x, "Duration"] > 120:
    df.drop(x, inplace = True)

#remember to include the 'inplace = True' argument to make the changes in the original DataFrame object instead of returning a copy

print(df.to_string())
```
